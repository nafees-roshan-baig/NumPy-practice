```python
import numpy as np
a = np.arange(8).reshape(2,4)
print("org array is:\n", a)
print("flatten array is:\n",a.flatten(order = 'F'))
print("flatten array is:\n",a.flatten(order = 'C'))
```

```python
#joining Arrays
#axis =0 --- row wise
#axis =1 --- column wise

a= np.array([[1,2],[3,4]])
b= np.array([[5,6],[7,8]])
print("jooining the two arrays along row wise:")
print(np.concatenate((a,b),axis=0))
print("jooining the two arrays along column wise:")
print(np.concatenate((a,b),axis=1))
```

```python
print(np.stack((a,b),axis=0))
print(np.stack((a,b),axis=0))
```

```python
print(np.vstack((a,b)))
print(np.hstack((a,b)))
```

```python
a = np.arange(9)
b=np.split(a,3)
print(b)
c=np.split(a,[4,7])
print(c)
```
